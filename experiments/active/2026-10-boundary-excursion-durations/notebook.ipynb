An earlier version of this cell said the calibrators never looked like this. **That was wrong in part:** the same diagnostic on the Rev4 calibration walkers (second printout below, relabelled eden→RLLR,
bal→RLLRL, rotor→RLLLRR, idla→RLLLLRLL) shows that Eden (longest τ ≈ 1.0 A) and IDLA (≈ 1.3 A, almost no populated mid-depth bins) also have excursions of order A.

The excursion population is **heavy-tailed, possibly bimodal**: the ants' weak upturn at d ≥ L (tens of excursions per window) is not matched by any restart-free walker, and whether it is real is open.
