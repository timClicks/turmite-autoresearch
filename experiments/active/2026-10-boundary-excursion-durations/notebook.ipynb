## Revision 3: time-weighted survival (calibration only, 2026-10-29)

**Question.** Number weighting gave every short boundary return (34–73% of excursions, <2% of time) a full vote. Does the *time-weighted* survival $S_{tw}(s)=\sum_{\tau>s}\tau/\sum\tau$ — the duration law of the excursion containing a random off-boundary step — collapse in $s/L^2$ for walkers whose excursions are diffusive, and in $s/L$ for mean-field?

**Assumptions.** Eden interior steps are exact SRW (all neighbours visited), so its excursions are SRW hitting times of the boundary set — D in theory, not by construction. Hence a synthetic gate (G1) with laws D/S by construction comes first. Same pairwise Score, margin 0.75, grid $2^{-8}..2^{8}$, Rev2 retention + $n_{eff}\ge 30$; censoring sensitivity adds the open excursion.

### What this does and does not show

- **Shows:** under frozen rules and fresh seeds, Score_tw classifies known-D/known-S synthetic laws and the compact D calibrators (Eden, IDLA) and the S calibrator (mean-field) correctly; the number-weighted replication reproduces Rev2 (eden S, idla inconclusive), so the old failures look like a Score bias.
- **Does not show anything about ants**: no ant data were run under Revision 3.
- **Caveat (post hoc):** the rotor also reads **D**, with a sharp cliff at $\tau\approx0.15$–$0.3\,A$. In shape-stable domains $A\propto L^2$, so Score_tw cannot separate diffusion from area-filling sweeps; it separates both from boundary-local sweeps ($\tau\propto L$). An ant D verdict must be read accordingly.
