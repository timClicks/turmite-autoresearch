# Question and why it matters
Does changing the colour-2 medium around an online-detected ray disrupt alternating-turn staircases and cause earlier geometric termination? This is the intervention follow-up to S1 in [the README](../README.md); cross-cutting questions are in [notes](../../../../notes/open-questions-hypotheses.md). Predictions were registered before measurement at 2026-10-15 21:18 +13:00, base da88e61.
## Assumptions and definitions
Sixteen causally selected states per rule share one blank-start trajectory; seed is not applicable. A radius-128 square is assumed to contain relevant medium, but only current colour-2 cells are edited. Sham 2→2, reset 2→0 and advance 2→3 preserve the ant and all other colours. Termination is first failure after step 128 of the original-heading rolling 128-step diagonal criterion, censored at 2048. It does not mean permanent termination. Full-window alternating-turn fraction includes recovery after the original ray ends.

## Results
A, B and C failed for every rule. Reset to 0 gave alternating-fraction drops ≥0.10 in 0/16, 0/16, 2/16 trials and earlier termination in 3/16, 6/16, 8/16. Mean full-window alternating fractions instead increased by 0.050501, 0.073156, 0.041891. Advance to 3 caused no earlier termination (0/16 each). All arms were uncensored by 2048. These results reject the registered strong disruption predictions for this local intervention. They do not exclude a medium-dependent mechanism or identify the cell that ends a ray.
## Caveats and controls
Synthetic alternating staircase and axial controls passed; a 32-case property test checks the edited dense loop against reference::Ant. Sham replays matched all 2048 steps and edited zero cells; the full release suite passed. These are deterministic within-trajectory comparisons, not independent seeded replicates; reported SD is between trials. All selected states are from early eligible detections, and the 128-step latency can hide immediate disruption. Paths can leave the edited square (counts in CSV).
Post-measurement interpretation: colours 2 and 3 have the same turn letter in every tested rule; 2→3 affects immediate colour phase but not the first turn. A change needs a revisit to affect motion. This limits C as a disruption control; it was not removed or redefined.
## What this changes / next steps
A simple claim that erasing colour-2 medium generically suppresses staircases is unsupported. Next distinguish first-entry effects from revisit effects and record exposure to edited cells. Open: do particular ray-ending defects, rather than the abundant ground, govern termination? Register any endpoint-targeted experiment separately.
