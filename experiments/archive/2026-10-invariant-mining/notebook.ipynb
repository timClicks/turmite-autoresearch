Question: which additive functions of position, heading, time and colours survive every local transition? See README for the induction argument and registered controls. No randomness or fit windows. These laws are necessary constraints, not an explanation of growth.


```python
from pathlib import Path
import hashlib, json
import polars as pl
root = Path.cwd().resolve()
while not (root / 'Cargo.toml').exists():
    if root == root.parent: raise RuntimeError('Run within the repository')
    root = root.parent
results = root / 'experiments/archive/2026-10-invariant-mining/results'
provenance = json.loads((results / 'provenance.json').read_text())
assert hashlib.sha256((results / 'invariants.csv').read_bytes()).hexdigest() == provenance['output_sha256']
print(json.dumps(provenance, indent=2))
```

```python
laws = pl.read_csv(results / 'invariants.csv')
print(laws.group_by(['rule', 'modulus']).len().sort(['rule', 'modulus']))
print(provenance['colour_shift_symmetries'])
```

Counts include dependent multiples and are not invariant-space dimensions. Colour-shift symmetries act on the entire initial field; reflection also transforms position and heading.
