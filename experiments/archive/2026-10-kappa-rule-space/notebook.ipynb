## W5 bulk fingerprint on an area-matched mature field (`w5_stats.txt`, registered 2026-10-15 before any code ran; `w5_run.log`, `w5_obs.csv`, `w5_rules.csv`)

```python
print((res/'w5_run.log').read_text()); print((res/'w5_stats.txt').read_text())
```

**Shows.** Run to 20 000 visited cells, freeze a depth mask (depth >= 20), record 2^18 steps (3772 ant runs). Of 11 declared bulk observables none reaches the registered held-out gain of 0.15 over ln A(22): best +0.040 (`msd64`, CI -0.009..+0.085), `ret32` +0.039, `I_turn` +0.001 (P5.1 **failed**, P5.2 held). The small gain is real (permutation p about 0.0005) and is one signal carried by `msd64`, `ret32`, `reuse64` (more local reuse, higher kappa). Those reproduce across seeds (ICC >= 0.94) but the registered primary `I_turn` does not (0.56): P5.3 **failed** for the primary set. Rotor MSD/L falls with lag (0.59 -> 0.47 -> 0.21), as it should.

**Does not show.** That transport at long lags is irrelevant: lag-4096 observables are confinement-limited at this A0 and unreproducible (ICC 0.09), so subdiffusive transport is untested. It also does not give a mechanism for the local-reuse association, and the pool lost five fast-growing, high-kappa rules (mean ln kappa +0.86) to the 4-seed rule. The positive control on c shows nothing (c is a boundary quantity), and the `pw`/`rw` controls have no bulk at all; the planted-signal check recovers 78% of a planted effect. The unregistered extras in the file (obs alone, joint model, grouped CV, sensitivities) were added after the primary table was seen.

## What the results show (from the README's own verdicts; see its Log for the full tables)
- **W6 pilot (40 bases, 398 mutant pairs, 6 seeds, one window k=26).** P6.1 held (per-base median early/late |d ln kappa| ratio 2.21, CI 1.59..3.10), P6.2 held (late letters 5.2x the scaled replicate null), P6.3 held, P6.6 held barely (usable 0.716). Effect of a letter decays smoothly with position (Spearman -0.92 of median |d| on j): the prefix is a weighting, not a switch.
- **W5.** P5.1 failed (best bulk observable adds +0.040 over ln A(22); bar 0.15), P5.2 held, P5.3 failed for the primary `I_turn` (0.56) while `msd64`, `msd512`, `ret32`, `reuse64` reproduce (ICC >= 0.94). A small shared signal (more local reuse, higher kappa, about a tenth of the unexplained variance) is real but not an explanation. Long-lag transport is untested, not excluded.
- **W3b / W4.** All six W3b predictions held; strict held-out R^2 0.58 (p = 6) and 0.74 (p = 8) on fresh rules, size-controlled gain +0.145 (p = 6) and +0.206 (p = 8, exploratory, singled out by the same data). W4: P4.2, P4.1 scaling, P4.5 held; P4.1 threshold, P4.3, P4.4 (by 0.005), P4.6 rotor band failed as registered; the pw control was not obtained.

## What it does not show
- **Size is not separated from the mutation result.** The early/late contrast in kappa equals that in ln A(22) (ratio 1.01) and |d ln kappa| tracks |d ln A| (rho +0.69), as expected at fixed t. The result says early letters change how fast the pattern grows, which is the growth-prefactor question, not a rule-level kappa independent of size. The sibling experiment `2026-10-kappa-matched-area` found that at matched area the residual rule effect is not established (noise-floor criteria failed).
- The p = 8 gain was a lead in W3b and replicated in W3c when the lookup is trained on stationary census rules (it nearly vanishes if trained on all census rules; see cross check). W6 is a pilot (bases are the unit, position confounded with letter adjacency and the base prefix). The README sentence claiming mutants were "unrelated to the census/pool fit" was wrong and is corrected in the Log: all mutants are census rules; what is new is the paired interventional design and fresh runs (W6 vs census ln kappa Spearman +0.93 on 292 of 299 usable rules).
- Selection may understate the early effect (early mutants escape more, 17% vs 8%); untested.
