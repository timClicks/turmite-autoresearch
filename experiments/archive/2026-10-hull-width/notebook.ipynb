# Hull width: is the ants' wide edge coastline roughness or porosity?

Experiment `2026-10-hull-width`. Written for a reader who has not seen the work. Links: `README.md`, `NOTEBOOK.md`, `notes/frontier-results.md` (P7).

**Question.** The surface-width W_surf of ants grows as ~t^0.22-0.25 against ~t^0.10-0.13 for the rotor walker. Surface cells include the rims of interior holes. Is the extra width (H-hole) porosity or (H-rough) the outer coastline itself?

**Assumptions.** Exterior = unvisited cells 4-connected to the grid border (diagonal channels count as closed; not checked). The angular profile r(theta) is the outermost cell per bin (extremal, misses overhangs). Cohort: six A4 rules x 4 seeds (221-224) and 8 rotor seeds, `MAX_LOG2 = 30`, 8192^2 grid. All exponents are finite-window (k >= 22).
Everything here was registered in the README before it was run; this notebook reads only committed `results/`.

```python
import hashlib, pathlib, subprocess
res = pathlib.Path('results')
if not res.exists(): res = pathlib.Path('experiments/archive/2026-10-hull-width/results')
print('git HEAD (notebook run):', subprocess.run(['git','rev-parse','--short','HEAD'],capture_output=True,text=True).stdout.strip())
for f in ['hull_ants.csv','hull_rotor.csv','shape_check.csv','hull_stats.txt','hull_bands.txt','hull.commit']:
    print(f'{f:18s}', hashlib.sha256((res/f).read_bytes()).hexdigest()[:16])
print('producing commit of the run:', (res/'hull.commit').read_text().strip()[:8])
```

## Registered analysis (`hull_stats.txt`)

```python
print((res/'hull_stats.txt').read_text())
```

**Shows.** Calibration held (rotor W_out exponent 0.135 in [0.08, 0.14]; W_all agrees with `shape` Wsurf to 4 digits on 160 rows). Ants' W_out exponent 0.239, 1.77x the rotor's, so the extra width is on the outer interface: **H-rough**, H-hole rejected (ants' hole share of the edge is only ~4% vs 25% for the rotor, so the registered prediction 2 failed in the direction opposite to my guess). Spectrum and box-count dimension agree on direction, but a larger spectral exponent means a *smoother* profile at small scale, so prediction 4 was met numerically with the wrong sign convention in my own wording.
**Does not show.** Where on the profile the extra width sits; that was an unregistered reading, tested below.

## Follow-up: ants/rotor band-power ratio (`hull_bands.txt`, registered 2026-10-15 before the script existed)

```python
print((res/'hull_bands.txt').read_text())
```

**Shows.** Relative to the rotor, the ants carry 2-6x the long-wavelength power (bands 3-4 to 17-32), decreasing monotonically with wavenumber (Spearman -1.00 / -0.96). Predictions 1, 3, 4 held; prediction 2 (deficit at 65-128 in >= 5/6 rules at both k) **failed** (R = 0.72 then 0.97; 5/6 then 3/6 rules); prediction 5 (rotor band power stable between k = 28 and 30) **failed** (x0.46-0.65 per band) but was mis-specified: the spectrum is divided by rbar^2, so it scales as (W/R)^2, which predicts x0.58 for the rotor and x0.77 for the ants from the measured W exponents (observed ~0.75 for the ants). The apparent 'ants do not decay like the rotor' lead only restates the W-exponent gap. Registered rule: mixed.
**Does not show.** A real-space structure function or lobe size (profiles were not saved); anything beyond the W-exponent gap about how band power changes with k (two checkpoints only).

## Caveats
6 rules chosen earlier, not census-sampled; 4 seeds per rule; rotor noise at the lowest band is large (run sd of log ~ 0.8-1.1); band power is in units of rbar^2 and rbar grows; hull dimension is over a short range s = 2..32; extremal profile.
