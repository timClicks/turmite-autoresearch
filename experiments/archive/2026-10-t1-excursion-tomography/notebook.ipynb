## Question and why it matters
Continue T1b: describe the extreme completed-excursion cutoff relative to radius. See README.md and ../../notes/excursion-tomography-2026-10-13.md. This is reuse of observational data, not independent confirmation.

## Assumptions and definitions
L999 uses log-linear interpolation in its containing log2 bin. Bounds span one octave. Escaped windows are excluded; unavailable bin records reduce seed counts. Open-time filtering cannot correct censoring. Patched seeds are replicates within rules.

## Provenance, data and method
Run scripts/t1b_cutoff.py under uv. Heavy simulation is historical; this notebook only reads results. New outputs await user-authorized commit. Input hashes are checked below.

```python
from pathlib import Path
import hashlib, json, subprocess
import polars as pl
SEED = 20261014
root = next(p for p in [Path.cwd(), *Path.cwd().parents] if (p / 'Cargo.toml').exists())
exp = root / 'experiments/archive/2026-10-t1-excursion-tomography'
summary = json.loads((exp / 'results/t1b_cutoff.json').read_text())
print('Analysis commit:', summary['commit'], 'seed:', SEED)
print('Reading date:', subprocess.check_output(['date', '-I'], text=True).strip())
for relative, expected in summary['inputs'].items():
    actual = hashlib.sha256((root / relative).read_bytes()).hexdigest()
    assert actual == expected
    print(relative, actual)
for name in ['t1b_cutoff.json', 't1b_cutoff_rules.csv', 't1b_cutoff_seeds.csv']:
    print(name, hashlib.sha256((exp / 'results' / name).read_bytes()).hexdigest())
print(json.dumps(summary['windows'], indent=2))
print('Controls:', summary['controls'])
rules = pl.read_csv(exp / 'results/t1b_cutoff_rules.csv')
print(rules.filter(pl.col('k') == 25).select('rule', 'n_seeds', 'ln_ratio_mean', 'ln_ratio_sd').head(8))
```

## Results
Negative association at both windows matches the descriptive prediction. This does not establish a mechanism; radius remains in the denominator.

## Caveats and controls
Between-seed sample sd is reported, not fit error. The quantile has octave resolution. Rotor baseline misses the registered 2.00 +/- 0.01 band; random-walk controls are unavailable at k=25. These controls prevent treating this analysis as calibrated mechanistic evidence.

## What this changes / next steps
The descriptive ratio association persists. Next: matched-area or intervention study, exact quantile resolution and censoring treatment. Do not infer that excursion duration is a spatial chord length.
