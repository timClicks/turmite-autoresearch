# Does rule-specific kappa survive controlling for territory size?

Experiment `2026-10-kappa-matched-area`. Written for a reader who has not seen the work. Observational reanalysis only: **no new simulation**.

**Question.** In the T1/T1b patched-start windows (44 rules, k = 22 and 25, 6-8 seeds each) kappa and territory area A are tied at fixed t by dA/dt = c kappa P/A. Once A is held fixed, is there still a rule-level kappa, and does it track the extreme-excursion cutoff ln(L_99.9/R_eq)?

**Assumptions.** Seeds are exchangeable replicates of a rule; windows from one seed at k=22 and 25 are not independent (the pooled ICC ignores this); escaped windows are excluded as in T1b; area adjustment is linear in ln A; matching uses rule-mean ln A with tolerance 0.05 (0.10 as sensitivity).

Three analyses, each registered in `README.md` *before* it was run: (1) `matched_area.py` (ICC, slopes, partial Spearman), (2) `matched_pairs.py` (matched-area pairs), (3) `noise_floor.py` (seed-noise floor). Notebook reads only committed `results/`.

## 2. Matched-area pairs (`matched_pairs.txt`)

```python
print((res/'matched_pairs.txt').read_text())
```

**Does not show:** that the residual exceeds seed noise; that is analysis 3.

## 3. Seed-noise floor (`noise_floor.txt`)

```python
print((res/'noise_floor.txt').read_text())
```

**Shows (registered tolerance 0.05):**

| | k=22 | k=25 | registered | verdict |
|---|---|---|---|---|
| N1 seed sd ln kappa / pair SE | 0.089 / 0.043 | 0.089 / 0.037 | <= 0.10 / <= 0.05 | held |
| N1 frac pairs z>2 | 0.39 | 0.58 | >= 0.60 | **failed** |
| N2 P-b on z>2 pairs | 0.92 [0.50, 1.00] (24 pairs) | 0.95 [0.78, 1.00] (39) | >= 0.75, interval excludes 0.5 | k=22 **failed**, k=25 held |
| N3 split-seed P-b | 0.79 [0.59, 0.95] | 0.81 [0.61, 0.95] | >= 0.65 | held |

By the registered decision rule N1 failed, so P-b is uninformative at this noise level and **no T2 is designed on this pair set**.
**Points the other way (not used to rewrite the rule):** N3 held, so the association is not within-trajectory coupling of kappa and ln_ratio; at k=25 the z <= 2 complement is at 0.50 (about what pure noise gives) while resolved pairs are at 0.95.

## Caveats
- z uses seed sd of ln kappa only; the z > 2 cut is a selection on |d ln kappa|, so the high P-b there is partly built in.
- Pairs share rules, so pair counts overstate independent information; the bootstrap is by rule, and the k=22 interval is wide.
- Matching on noisy mean ln A can leak area noise into kappa through the identity above.
- k=22 and k=25 windows share seeds and rules; they are not independent replications.
- 44 rules, 6-8 seeds each.

## Conclusion and next steps
Rule-specific kappa at fixed area is **not established**: the effect may be real and small (N3, k=25 complement) but the matched pairs cannot resolve it in about half the cases, and two of the three registered noise-floor criteria failed. The paper's rule-ICC for kappa should be qualified as "at fixed t".
